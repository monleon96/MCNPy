# PlotBuilder Legend Tests

Visual verification of legend `ncol` and `loc='outside ...'` features.

```python
import numpy as np
import matplotlib.pyplot as plt
from kika.plotting import PlotBuilder, PlotData

# Helper: generate several labeled curves
def make_curves(n=5):
    x = np.linspace(0, 2 * np.pi, 100)
    curves = []
    for i in range(n):
        y = np.sin(x + i * 0.4) * (1 - 0.1 * i)
        curves.append(PlotData(x=x, y=y, label=f'Series {i+1}'))
    return curves
```

## 1. Inside legend — default (1 column) vs 2 columns

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for curve in make_curves():
    b1 = PlotBuilder(ax=axes[0])
    # first pass — just add data to ax[0]
    break  # setup below

# Left: default ncol (1 column)
b1 = PlotBuilder(ax=axes[0])
for c in make_curves():
    b1.add_data(c)
(b1.set_labels(title='Inside — ncol=1 (default)', x_label='x', y_label='y')
   .set_legend(loc='upper right')
   .build())

# Right: ncol=2
b2 = PlotBuilder(ax=axes[1])
for c in make_curves():
    b2.add_data(c)
(b2.set_labels(title='Inside — ncol=2', x_label='x', y_label='y')
   .set_legend(loc='upper right', ncol=2)
   .build())

fig.tight_layout()
plt.show()
```

## 2. Inside legend — ncol=3 with many series

```python
b = PlotBuilder(figsize=(10, 6))
for c in make_curves(9):
    b.add_data(c)
(b.set_labels(title='Inside — ncol=3, 9 series', x_label='x', y_label='y')
  .set_legend(loc='lower center', ncol=3)
  .build())
plt.show()
```

## 3. Outside legend — all three directions

```python
positions = ['outside right', 'outside top', 'outside bottom']

for pos in positions:
    b = PlotBuilder(figsize=(9, 5))
    for c in make_curves():
        b.add_data(c)
    (b.set_labels(title=f"loc='{pos}'", x_label='x', y_label='y')
      .set_legend(loc=pos)
      .build())
    plt.show()
```

## 4. Outside legend with multiple columns

```python
b = PlotBuilder(figsize=(10, 6))
for c in make_curves(8):
    b.add_data(c)
(b.set_labels(title="outside bottom, ncol=4", x_label='x', y_label='y')
  .set_legend(loc='outside bottom', ncol=4)
  .build())
plt.show()
```

```python
b = PlotBuilder(figsize=(10, 6))
for c in make_curves(6):
    b.add_data(c)
(b.set_labels(title="outside right, ncol=2", x_label='x', y_label='y')
  .set_legend(loc='outside right', ncol=2)
  .build())
plt.show()
```

## 5. Different figure sizes with outside legend

```python
for w, h in [(7, 4), (12, 4), (8, 8)]:
    b = PlotBuilder(figsize=(w, h))
    for c in make_curves():
        b.add_data(c)
    (b.set_labels(title=f'figsize=({w}, {h}), outside right', x_label='x', y_label='y')
      .set_legend(loc='outside right')
      .build())
    plt.show()
```

## 6. Font sizes are independent (labels vs ticks vs legend)

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Left: large labels, small ticks
b1 = PlotBuilder(ax=axes[0])
for c in make_curves(3):
    b1.add_data(c)
(b1.set_labels(title='Large labels, small ticks', x_label='x', y_label='y')
   .set_legend(loc='upper right')
   .set_font_sizes(title=20, labels=18, ticks=8, legend=14)
   .build())

# Right: small labels, large ticks
b2 = PlotBuilder(ax=axes[1])
for c in make_curves(3):
    b2.add_data(c)
(b2.set_labels(title='Small labels, large ticks', x_label='x', y_label='y')
   .set_legend(loc='upper right')
   .set_font_sizes(title=10, labels=8, ticks=18, legend=7)
   .build())

fig.tight_layout()
plt.show()
```

## 7. Outside top — no overlap with title (various font sizes)

```python
for title_size in [14, 20, 28]:
    b = PlotBuilder(figsize=(10, 5))
    for c in make_curves(6):
        b.add_data(c)
    (b.set_labels(title=f'Title fontsize={title_size}', x_label='x', y_label='y')
      .set_font_sizes(title=title_size)
      .set_legend(loc='outside top', ncol=3)
      .build())
    plt.show()
```

## 8. Outside bottom — no overlap with x-label (various font sizes)

```python
for label_size in [12, 18, 24]:
    b = PlotBuilder(figsize=(10, 5))
    for c in make_curves(6):
        b.add_data(c)
    (b.set_labels(title='Bottom legend test', x_label=f'X-axis (fontsize={label_size})', y_label='y')
      .set_font_sizes(labels=label_size)
      .set_legend(loc='outside bottom', ncol=3)
      .build())
    plt.show()
```

## 9. Outside top with no title — legend should still work

```python
b = PlotBuilder(figsize=(10, 5))
for c in make_curves():
    b.add_data(c)
(b.set_labels(x_label='x', y_label='y')
  .set_legend(loc='outside top', ncol=5)
  .build())
plt.show()
```
